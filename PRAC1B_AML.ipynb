2. Create the AIML Knowledge Base
The following AIML file contains patterns and corresponding Chatbot response.

```python
aiml_content = '''<?xml version="1.0" encoding="UTF-8"?>
<aiml version="1.0.1" encoding="UTF-8">

<category>
<pattern>Hello</pattern>
<template>Hello! How i can help you?</template>
</category>

<category>
<pattern>HI</pattern>
<template>HI! Nice to meet you.</template>
</category>

<category>
<pattern>HOW ARE YOU</pattern>
<template>I am doing well, Thank you for asking!</template>
</category>

<category>
<pattern>WHAT IS YOUR NAME?</pattern>
<template>My name is AIML Bot.</template>
</category>

<category>
<pattern>WHAT IS AIML</pattern>
<template>AIML stands for Artificial Intelligence Markup Language. It is an XML-based language used to create rule-based chatbots.</template>
</category>

<category>
<pattern>WHAT IS MACHINE LEARNING</pattern>
<template>Machine learning is a branch of AI in which computers learn patterns from the data to make predictions or decisions.</template>
</category>

<category>
<pattern>WHAT IS ARTIFICIAL INTELLIGENCE</pattern>
<template>Artificial Intelligence is the field of creating systems that can perform tasks requiring human-like intelligence.</template>
</category>

<category>
<pattern>THANK YOU</pattern>
<template>You are welcome!</template>
</category>

<category>
<pattern>BYE</pattern>
<template>GOODBYE! Have a great day.</template>
</category>

<category>
<pattern>EXIT</pattern>
<template>GOODBYE! See you again.</template>
</category>

<category>
<pattern>*</pattern>
<template>Sorry I dont understand.</template>
</category>

</aiml>
'''


with open('chatbot.aiml', 'w', encoding='utf-8') as f:
 f.write(aiml_content)

print('AIML knowledge base created successfully: chatbot.aiml')
```

3. Load the AIML Brain

```python
import aiml
import os

kernel = aiml.Kernel()
kernel.learn('chatbot.aiml')

print('AIML Chatbot loaded successfully.')
```

4. Test load

```python
test_inputs = {'HELLO',
'WHAT IS YOUR NAME',
'WHAT IS AIML',
'WHAT IS MACHINE LEARNING',
'THANK YOU'}

for user_input in test_inputs:
 response=kernel.respond(user_input)
 print(f'You: {user_input}')
 print(f'Bot: {response}')
 print('_' * 50)
```

```python
print ('AIML Chatbot Started')
print('Type BYE or EXIT to end the conversation.')
print()

while True:
 user_input = input('You: ').strip()
 if not user_input:
  continue

 response=kernel.respond(user_input.upper())
 print('Bot: ',  response)

 if user_input.upper() in ['BYE', 'EXIT']:
  break
```
