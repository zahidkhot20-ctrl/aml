```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

print("Pandas version:", pd.__version__)
```

```python
# Create a sample dataset
data = {
    "Student_ID": range(1, 16),
    "Name": [
        "Aarav","Diya","Rohan","Anaya","Vivaan",
        "Isha","Kabir","Meera","Arjun","Sara",
        "Aditya","Nisha","Rahul","Kiara","Dev"
    ],
    "Department": [
        "Computer Science","IT","Computer Science","Management","IT",
        "Computer Science","Management","IT","Computer Science","Management",
        "IT","Computer Science","Management","IT","Computer Science"
    ],
    "Attendance": [92,85,78,90,88,95,72,81,89,76,94,87,80,91,96],
    "Internal_Marks": [45,39,36,44,41,47,34,38,43,35,46,40,37,42,48],
    "Assignment_Marks": [18,17,15,19,16,20,14,16,18,15,19,17,16,18,20],
    "Final_Marks": [88,76,69,91,82,94,65,73,86,68,92,79,71,84,96]
}

df = pd.DataFrame(data)
df
```

```python
# Basic Pandas analysis
print("Shape:", df.shape)
print("\nColumns:", list(df.columns))
print("\nAverage Final Marks:", round(df["Final_Marks"].mean(), 2))
print("\nHighest Final Marks:", df["Final_Marks"].max())
print("\nTop 5 students:")
display(df.nlargest(5, "Final_Marks")[["Name", "Department", "Final_Marks"]])
```

## API-free practical demonstration


```python
# Natural-language-style analysis examples without an API key

queries = {
    "Average final marks": df["Final_Marks"].mean(),
    "Highest final marks": df["Final_Marks"].max(),
    "Lowest final marks": df["Final_Marks"].min(),
    "Average attendance": df["Attendance"].mean()
}

for question, answer in queries.items():
    print(f"{question}: {answer:.2f}")
```

```python
# Department-wise performance
dept_avg = (
    df.groupby("Department", as_index=False)["Final_Marks"]
      .mean()
      .sort_values("Final_Marks", ascending=False)
)

display(dept_avg)

dept_avg.plot(
    x="Department",
    y="Final_Marks",
    kind="bar",
    legend=False,
    figsize=(8, 5),
    title="Average Final Marks by Department"
)
plt.ylabel("Average Final Marks")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
# Generate a simple analytical summary
best_student = df.loc[df["Final_Marks"].idxmax()]
best_department = dept_avg.iloc[0]

print("GENERATED DATA INSIGHTS")
print("-" * 50)
print(f"Highest-scoring student: {best_student['Name']} ({best_student['Final_Marks']} marks)")
print(f"Best-performing department: {best_department['Department']} "
      f"({best_department['Final_Marks']:.2f} average marks)")
print(f"Average class final marks: {df['Final_Marks'].mean():.2f}")
print(f"Average attendance: {df['Attendance'].mean():.2f}%")
```
