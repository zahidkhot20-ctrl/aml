```python
environment={
    'A':"Dirty",
    'B':"Dirty",

}
agent_location='A'
print('Initial environmrnt:',environment)
print('agent location:',agent_location)
class VacumCleanerAgent:
    def __init__(self,location="A"):
        self.location=location
        self.actions=0
        self.cleaned_rooms=0

    def perceive(self,environment):
        "Sense the condition of the current room."
        return environment[self.location]

    def decide_actions(self,percept):
      "Choose an action based on the current percept."
      if percept=="Dirty":
        return "Clean"
      elif self.location=="A":
        return "Move right"
      else:
        return "Move left"

    def act(self,action,environment):
      "Execute the action."
      self.actions +=1

      if action=="Clean":
        environment[self.location]="Cleaned"
        self.cleaned_rooms+=1
      elif action == 'move right':
        self.loaction = 'B'
      elif action == 'move left':
        self.location ='A'
```

4.Run the Intelligent Agent

```python
agent = VacumCleanerAgent(location='A')

print('--- Intelligent Agent Execution ---')

max_steps = 18

for step in range(1, max_steps + 1):
  percept = agent.perceive(environment)
  action = agent.decide_actions(percept)
  agent.act(action, environment)

  print(f'Step {step}:')
  print(f'  Location: {agent.location}')
  print(f'  Percept: {percept}')
  print(f'  Action: {action}')
  print(f'  Rooms: {environment}')
  print()

  if all(status == 'Clean' for status in environment.values()):
    print('Goal achieved: All rooms are clean.')
    break
```

5.Performance Evalution

```python
clean_rooms = sum(1 for status in environment.values() if status == 'Clean')
total_rooms = len(environment)
cleaning_percentage = (clean_rooms / total_rooms) * 100

print('--- Performance Evaluation ---')
print('Total Rooms:', total_rooms)
print('Clean Rooms:', clean_rooms)
print('Cleaning Rate:', f'{cleaning_percentage:.2f}%')
print('Actions Performed:', agent.actions)
print('Agent Location:', agent.location)

```

6.Interactive Intelligent Agent

```python
def get_room_status(room):
  while True:
    value = input(f'Enter status of Room {room} (Clean/Dirty): ').strip().capitalize()
    if value in ['Clean', 'Dirty']:
      return value
    print('Please enter either Clean or Dirty.')

interactive_environment = {
  'A': get_room_status('A'),
  'B': get_room_status('B'),
}

interactive_agent = VacumCleanerAgent(location='A')

print('\nInitial Environment:', interactive_environment)

for step in range(1, 11):
  percept = interactive_agent.perceive(interactive_environment)
  action = interactive_agent.decide_actions(percept)
  interactive_agent.act(action, interactive_environment)

  print(f'Step {step}: Percept={percept}, Action={action}, Environment={interactive_environment}')

  if all(status == 'Clean' for status in interactive_environment.values()):
    print('Goal achieved: All rooms are clean.')
    break
```
